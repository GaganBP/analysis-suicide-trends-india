```python
# Train model
model_state = LinearRegression()
model_state.fit(state_df[['2021 Total']], state_df['2021 Total'])

# Predict with growth factor
growth_factor = 1.1 ** 3  # 10% annual growth for 3 years
state_predictions = pd.DataFrame({
    'State': state_df['State'],
    '2021 Original': state_df['2021 Total'],
    '2024 Predicted': model_state.predict(state_df[['2021 Total']]) * growth_factor
})

# Clean output
state_predictions['2024 Predicted'] = state_predictions['2024 Predicted'].clip(lower=0).round(0)

# Print without converting State names to integers
print("\nOriginal and Predicted Total Suicides for States:")
print(state_predictions)  # Let pandas handle mixed types automatically
```

```python
# First train the model for Union Territories
ut_model = LinearRegression()
ut_model.fit(ut_df[['2021 Total']], ut_df['2021 Total'])

# Predict with growth factor (same 10% annual growth assumption)
growth_factor = 1.1 ** 3
ut_predictions = pd.DataFrame({
    'Union Territory': ut_df['Union Territory'],
    '2021 Original': ut_df['2021 Total'],
    '2024 Predicted': ut_model.predict(ut_df[['2021 Total']]) * growth_factor
})

# Clean and format output
ut_predictions['2024 Predicted'] = ut_predictions['2024 Predicted'].clip(lower=0).round(0)

# Add serial numbers
ut_predictions.insert(0, 'Sl.', range(1, len(ut_predictions) + 1))
ut_predictions.set_index('Sl.', inplace=True)

print("\nOriginal and Predicted Total Suicides for Union Territories:")
print(ut_predictions)
```

```python
# First train the model for Cities
city_model = LinearRegression()
city_model.fit(city_df[['2021 Total']], city_df['2021 Total'])

# Predict with growth factor (same 10% annual growth assumption)
growth_factor = 1.1 ** 3  # 3 years of 10% annual growth

# Create predictions DataFrame
city_predictions = pd.DataFrame({
    'Sl.': range(1, len(city_df) + 1),
    'City': city_df['Cities'].str.replace(r'\(.*\)', '', regex=True),  # Remove anything within parentheses
    '2021 Original': city_df['2021 Total'],
    '2024 Predicted': city_model.predict(city_df[['2021 Total']]) * growth_factor
}).set_index('Sl.')

# Clean and format output
city_predictions['2024 Predicted'] = city_predictions['2024 Predicted'].clip(lower=0).round(0)

print("\nOriginal and Predicted Total Suicides for Cities:")
print(city_predictions)
```
